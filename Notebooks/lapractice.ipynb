## Imports

```python
# necessary imports
import pandas as pd
import numpy as np
import os
import re
```

# Method for adding students attributtes

```python
def add_student(student_list):
    """Adds a new student to the list"""
    id = input("Enter ID: ")
    name = input("Enter Name: ")
    
    try:
        age = int(input("Enter Age: "))
    except ValueError:
        print("Invalid age. Must be a number.")
        return

    courses = input("Enter Courses (comma separated): ").split(",")

    student = {
        "id": id,
        "name": name,
        "age": age,
        "courses": [c.strip() for c in courses]
    }

    student_list.append(student)
    print("Student added successfully!")
```

# viewing students

```python
def view_students(student_list):
    """Displays all students"""
    if not student_list:
        print("No students found.")
        return

    for s in student_list:
        print("\nID:", s["id"])
        print("Name:", s["name"])
        print("Age:", s["age"])
        print("Courses:", ", ".join(s["courses"]))
```

# searching student

```python
def search_student(student_list, student_id):
    """Searches for a student by ID"""
    for s in student_list:
        if s["id"] == student_id:
            print("\nStudent Found:")
            print(s)
            return
    print("Student not found.")
```

# for courses

```python
def get_unique_courses(student_list):
    """Returns a set of unique courses"""
    courses = set()
    for s in student_list:
        courses.update(s["courses"])
    return courses
```

# removing a student

```python
def remove_student(student_list, student_id):
    """Removes a student by ID"""
    for s in student_list:
        if s["id"] == student_id:
            student_list.remove(s)
            print("Student removed successfully!")
            return
    print("Student not found.")
```
