```python
# line plot

#import matplotlib, pandas and numpy
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# set the data for x and y axis
x = np.linspace(1, 11, 100)
y = x ** 2
y2 = x ** 3

# Create a plot using Matplotlib
plt.plot(x, y, color="red")
plt.plot(x, y2, color="blue")
# setting the labels for x and y axis
plt.xlabel('x')
plt.ylabel('sin(x)')

# set the title of the plot
plt.title('Plot of sin(x)')

# set to show the grid
plt.grid(True)

# display the plot
plt.show()
```

```python
# multiple plots

# set the data for x and y axis
x = np.linspace(0, 10, 100)
y = np.sin(x)
y2 = np.cos(x)

# Create a plot using Matplotlib
plt.plot(x, y, color="red", label="sin x")
plt.plot(x, y2, color="green", label="cos x")

# setting the labels for x and y axis
plt.xlabel('x')
plt.ylabel('sin(x) and cos(x)')

# set the title of the plot
plt.title('Cos ans sin plot')

# set to show the grid
plt.grid(True)

# show legend
plt.legend()

# display the plot
plt.show()
```

## 3D plots 

```python
from mpl_toolkits.mplot3d import Axes3D

# Create a figure and a 3D axis
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Generate data for 3D surface plot
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# Create a 3D surface plot
surf = ax.plot_surface(x, y, z, cmap='viridis')

# Add color bar which maps values to colors
fig.colorbar(surf)

# Label axes
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

# Set title
ax.set_title('3D Surface Plot')

# Show the plot
plt.show()
```

### Exporting plots

<p>You can easily export or save plots created with Matplotlib in different formats (e.g., PNG, PDF, SVG, etc.) using the savefig() function. After creating your plot, just call plt.savefig() to save the figure to a file.</p>

```python
import seaborn as sns
df = sns.load_dataset('flights')
y1 = df[df['year'] == 1949]['passengers']
y2 = df[df['year'] == 1950]['passengers']

x = df[df['year'] == 1949]['month']
plt.plot(x, y1, label="1949 data")
plt.plot(x, y2, color="red", label="1950 data")
plt.legend()
plt.title("1949 vs 1950 data")
plt.grid(True)


# save the plot
plt.savefig('yearly_sales.png')
plt.show()
```

#### Parameters

* fname- The file name or path to save the figure (.png, .pdf, .jpg, .svg).
* dpi- Controls the resolution of the output image. A higher dpi results in a clearer image.
* bbox_inches- Ensures the saved image has no extra white space around the plot (none, titght etc).
* format- Explicitly specify the format, by default, it's inferred from the file extension.

```python
import seaborn as sns
df = sns.load_dataset('flights')
y1 = df[df['year'] == 1949]['passengers']
y2 = df[df['year'] == 1950]['passengers']

x = df[df['year'] == 1949]['month']
plt.plot(x, y1, label="1949 data")
plt.plot(x, y2, color="red", label="1950 data")
plt.legend()
plt.title("1949 vs 1950 data")
plt.grid(True)


# save the plot
plt.savefig(fname='yearly_sales.png', dpi=1000, bbox_inches="tight")
```

## Types of plots

<p>Seaborn offers a rich set of plot types specifically designed for statistical data visualization.</p>

### Distribution Plots
- **Histogram:** Visualizes the frequency distribution of a numeric variable using bars.
- **Kernel Density Estimation (KDE) plot:** Represents the probability distribution of a numeric variable using a smooth curve.
- **Rug Plot:** Shows individual data points along an axis, often used in conjunction with histograms.
- **Empirical Cumulative Distribution Function (ECDF) plot:** Shows the probability that a data point is less than or equal to a certain value.
### Relational Plots
- **Scatter Plot:** Visualizes the relationship between two numeric variables using points.
- **Line Plot:** Shows trends over time or another numeric variable using connected line segments.
- **Joint Plot:** Combines a scatter plot with a histogram or KDE plot on the margins to visualize both the relationship and distributions of two variables.
### Categorical Plots
- **Bar Plot:** Compares groups or categories using rectangular bars representing their counts or means.
- **Box Plot:** Shows the distribution of a numeric variable within different categories using boxes and whiskers.
- **Violin Plot:** Similar to a box plot, but also shows the kernel density of the data within each category.
- **Strip Plot:** Displays individual data points along an axis, categorized by another variable.
- **Swarm Plot:** Similar to a strip plot, but jitters the points horizontally to avoid overplotting.
### Correlation Analysis
- **Heatmap:** Visualizes the correlation matrix of a dataset using colored squares, where the color intensity represents the correlation coefficient between variables. 
- **Time Series Plots:** Line plots are often used for time series data, where the x-axis represents time. Additional functionalities in Seaborn can be used for time series analysis.

## Displot

* A function used to create distribution plots, which are visualizations that depict how your data is spread out.
* It offers a convenient way to explore the distribution of a single variable in your dataset.
Displot provides a high-level interface for creating various distribution plot types within Seaborn, including:
- **Histograms**
<p>Bar charts representing the frequency of data points within specific ranges (bins).</p>
- **Kernel Density Estimates (KDE plots)**
<p>Smooth curves that depict the underlying probability distribution of your data.</p>
- **Empirical Cumulative Distribution Functions (ECDF plots)**
<p>Plots showing the probability that a data point is less than or equal to a certain value.</p>

### Options
**Kind**
- kind="hist" (default): Creates a histogram.
- kind="kde": Creates a kernel density plot.
- kind="ecdf": Creates an ECDF plot.

**Customisations**
- hue: To represent distributions of different categories within the same plot.
- rug: To add a rug plot at the bottom, showing individual data points.
- color: To set the color of the plot elements.
- stat: To display additional statistics on the plot (e.g., mean, median). for hist ("count", "density", "percent", "probability", "proportion", "frequency")

### Pair plot

- Pairplot is a powerful function used to create a matrix of scatter plots, visualizing pairwise relationships between **all numeric** variables in your data.
- It's a valuable tool for data exploration, allowing you to identify correlations, patterns, and potential outliers across multiple variables simultaneously.
### Features
- **Pairwise Scatter Plots**
<p>pairplot generates a grid of scatter plots, where each row and column represent a variable in your data. Each plot displays the relationship between two variables, providing a comprehensive overview of pairwise interactions.</p>
- **Diagonal Plots**
<p>The diagonal plots within the grid can be customized to show the univariate distribution of each variable using histograms, kernel density estimates (KDE plots), or other distribution plots.</p>
- **Flexibility**
<p>Pairplot offers options to:<p/>
- Include only specific variables using the vars parameter.
- Use a different **hue** variable to represent categories within the scatter plots.
- Customize the diagonal plots using the **diag_kind** parameter.

### Histogram

<p>A histogram  is a visualization tool used to represent the distribution of a continuous variable. It depicts the number of data points that fall within specific ranges (bins) of the variable.</p>

**Understanding Histograms**
<p>A histogram is a bar chart where the x-axis represents the variable you're analyzing (e.g., height, weight, income), and the y-axis represents the frequency or density of data points within each bin (range or interval) on the x-axis.
The height of each bar corresponds to the number of data points that fall within that particular bin.</p>

**Optional Customization**
- bins: Specify the number of bins to use (e.g., bins=10 for 10 bins).
- stat: Calculate and display additional statistics on the plot (e.g., stat="mean").
- kde: Add a kernel density plot on top of the histogram to visualize the underlying probability distribution (e.g., kde=True).
- element: Choose how to represent the data (e.g., "bars" for bars, "step" for a step plot).
- color: Set the color of the histogram bars.

### Heat Maps
<p>Heatmaps are a graphical representation of data using a color matrix. They are particularly useful for visualizing relationships or patterns within a matrix of data, where the intensity of a color at a specific location represents the magnitude of the value at that point in the data. </p>
                                                                                                                                                                       
### Customisation
- **Data:** You can provide a DataFrame, a NumPy array, or a correlation matrix.
- **Colormap:** Specify the colormap using cmap (e.g., 'viridis', 'coolwarm').
- **Center:** Set the central reference point for the colormap using center. This is useful for bipolar data with both positive and negative values.
- **vmin, vmax:** Define the minimum and maximum values for the colormap using vmin and vmax.
- **Annot:** Include data values within each square using annot=True.
- **cbar:** Include a colorbar to show the colormap legend using cbar=True.
